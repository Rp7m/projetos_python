```python
#Importando o pacote pandas

import pandas as pd

#Importando o matplotlib para trabalhar com graficos
import matplotlib.pyplot as plt
```

```python
#Importando os dados em csv

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/dados/bebidas.csv")
```

```python
#Visualizando as primeiras 8 linhas do csv

df.head(8)
```

```python
#Realizando cópia dos dados para realizar alterações
dados = df.copy()
```

```python
#exibindo tabela dados cópia da tabela df
dados.head()
```

```python
#alterando nome das colunas
dados.rename(columns={'country':'País','beer_servings':'Porções_cervejas','spirit_servings':'bebida_espirituosa','wine_servings':'Porções_vinhos','total_litres_of_pure_alcohol':'litros_totais_de_álcool_puro'},inplace=True)
dados.head()
```

```python
#Exibindo informaçoes sobre a tabela
dados.info()
```

```python
#Exibindo os tipos de dados
dados.dtypes
```

```python
#Função para somar o numero de dados nulos combinando o isnull() com sum()
dados.isnull().sum()
```

```python
#Exibindo estatística numericas das colunas como os quartis
dados.describe()
```

```python
#Exibindo numero de linhas e colunas
dados.shape
```

```python
#Criando coluna total de porções de bebidas e atribuindo o valor total da soma das porçoes de bebidas
dados['Total_porcões'] = dados['Porções_cervejas'] + dados['bebida_espirituosa'] + dados['Porções_vinhos']
dados.head()
```

```python
#Renomeando coluna Total_porcões para Total_porções
dados.rename(columns={'Total_porcões':'Total_porções'},inplace=True)
dados.head()
```

```python
#Lista os 20 Paises que menos consomem bebidas álcoolicas
dados.groupby("País")["Total_porções"].min().sort_values(ascending=True).head(20)
```

```python
#Mostrando no grafico de barras
dados.groupby("País")["Total_porções"].min().sort_values(ascending=True).head(20).plot(kind='bar',title='Consumo de Bebidas por Países');
```

```python
#Lista os 10 Paises que mais consomem bebidas álcoolicas
dados.groupby("País")["Total_porções"].max().sort_values(ascending=False).head(10)
```

```python
#Mostrando no grafico
dados.groupby("País")["Total_porções"].max().sort_values(ascending=False).head(10).plot(kind='bar',title='Consumo de Bebidas por Países')
plt.ylabel('Total porções de álcool')
plt.xlabel('Países')
plt.show()
```
